```python
# 1a) Creating a dictionary of products indexed by product name
# Each product contains its ID, price per unit, weight per unit and quantity

products = {
       "Electric Kettle": {"ID":"ek", "Price_per_unit": 39.99, "Weight_unit": 1.20, "Qty": 47},
             "Desk Lamp": {"ID":"dl", "Price_per_unit": 24.50, "Weight_unit": 0.85, "Qty": 126},
      "Wireless Speaker": {"ID":"ws", "Price_per_unit": 59.99, "Weight_unit": 1.45, "Qty": 38},
              "Yoga Mat": {"ID":"ym", "Price_per_unit": 22.75, "Weight_unit": 1.10, "Qty": 84},
          "Water Filter": {"ID":"wf", "Price_per_unit": 31.20, "Weight_unit": 0.65, "Qty": 73},
   "Portable Power Bank": {"ID":"pb", "Price_per_unit": 44.95, "Weight_unit": 0.42, "Qty": 91},
       "Travel Backpack": {"ID":"tb", "Price_per_unit": 68.50, "Weight_unit": 0.95, "Qty": 32},
          "Food Storage": {"ID":"fs", "Price_per_unit": 18.99, "Weight_unit": 0.75, "Qty": 156},
    "Digital Desk Clock": {"ID":"dc", "Price_per_unit": 27.49, "Weight_unit": 0.55, "Qty": 64},
"Insulated Water Bottle": {"ID":"wb", "Price_per_unit": 29.95, "Weight_unit": 0.38, "Qty": 113}
           }

# 1b) Function to alter the price of any product

def change_price(product_name, new_price):
    """
    Changes the price of a product in the catalogue.

    Keyword arguments:
    product_name -- name of the product, e.g. "Desk Lamp"
    new_price -- new price in pounds; must be between £0.01 and £1000
    """
    if product_name not in products:
        print("Error: {} is not in the catalogue.".format(product_name))
        return False

    if new_price <= 0 or new_price > 1000:
        print("Error: price must be between £0.01 and £1000.")
        return False

    products[product_name]["Price_per_unit"] = round(new_price, 2)
    print("The price of {} is now £{:.2f}.".format(product_name, new_price))
    return True

    
# 1c) Function to alter the quantity of a product

def change_quantity(product_name, new_quantity):
    """
    Changes the quantity of a product held in inventory.

    Keyword arguments:
    product_name -- name of the product, e.g. "Yoga Mat"
    new_quantity -- new quantity; must be a whole number between 0 and 10000
    """
    if product_name not in products:
        print("Error: {} is not in the catalogue.".format(product_name))
        return False

    if not isinstance(new_quantity, int):
        print("Error: quantity must be a whole number.")
        return False

    if new_quantity < 0 or new_quantity > 10000:
        print("Error: quantity must be between 0 and 10000.")
        return False

    products[product_name]["Qty"] = new_quantity
    print("The quantity of {} is now {}.".format(product_name, new_quantity))
    return True

    
# 1d) Function to sort product names by quantity, price or weight

def sort_products(sort_by):
    """
    Returns a list of product names sorted by quantity, price or weight.

    Keyword arguments:
    sort_by -- "Qty", "Price_per_unit" or "Weight_unit"
    """
    valid_options = ["Qty", "Price_per_unit", "Weight_unit"]

    if sort_by not in valid_options:
        print("Error: sort_by must be Qty, Price_per_unit or Weight_unit.")
        return []

    sorted_products = sorted(products, key=lambda name: products[name][sort_by])
    return sorted_products

    
# 1e)i) Function to check the shopping cart

def check_cart(cart):
    """
    Checks that every product in the cart exists in the catalogue and that
    there is sufficient quantity available.

    Keyword arguments:
    cart -- a list of tuples, e.g. [("Desk Lamp", 2), ("Water Filter", 4)]

    Returns:
    A usable cart list containing only valid products with sufficient stock.
    """
    checked_cart = []

    for item in cart:
        product_name = item[0]
        quantity = item[1]

        if product_name not in products:
            print("Warning: {} is not in the catalogue and has been removed.".format(product_name))

        elif quantity <= 0:
            print("Warning: quantity for {} must be positive and has been removed.".format(product_name))

        elif quantity > products[product_name]["Qty"]:
            print("Warning: only {} {} available. Request for {} removed.".format(
                products[product_name]["Qty"], product_name, quantity))

        else:
            checked_cart.append((product_name, quantity))

    return checked_cart

    
# 1e)ii) Function to calculate cart costs

def calculate_cart_costs(cart):
    """
    Calculates the total sale price, 20% tax and shipping cost for a valid cart.

    Keyword arguments:
    cart -- a valid shopping cart list produced by check_cart()

    Returns:
    A dictionary containing subtotal, tax, shipping and total cost.
    """
    subtotal = 0.0
    total_weight = 0.0

    for product_name, quantity in cart:
        subtotal += products[product_name]["Price_per_unit"] * quantity
        total_weight += products[product_name]["Weight_unit"] * quantity

    tax = subtotal * 0.20
    shipping = total_weight * 0.85
    total = subtotal + tax + shipping

    costs = {
        "Subtotal": round(subtotal, 2),
        "Tax": round(tax, 2),
        "Shipping": round(shipping, 2),
        "Total": round(total, 2)
    }

    return costs

    
# 1e)iii) Function to print an invoice

def print_invoice(cart, costs):
    """
    Prints a simple invoice for the customer.

    Keyword arguments:
    cart -- the valid shopping cart list
    costs -- the dictionary of costs returned by calculate_cart_costs()
    """
    print("\n============= INVOICE ============")

    for product_name, quantity in cart:
        price = products[product_name]["Price_per_unit"]
        line_cost = price * quantity

        print("{:<25} {:>3} x £{:.2f} = £{:.2f}".format(
            product_name, quantity, price, line_cost))

    print("-----------------------------------------")
    print("Subtotal:                £{:.2f}".format(costs["Subtotal"]))
    print("Tax (20%):               £{:.2f}".format(costs["Tax"]))
    print("Shipping:                £{:.2f}".format(costs["Shipping"]))
    print("TOTAL:                   £{:.2f}".format(costs["Total"]))
    print("====================================\n")

  
# Test Cases

# b) Change a product price
change_price("Desk Lamp", 29.99)

# Invalid price test
change_price("Desk Lamp", -5)

# c) Change a product quantity
change_quantity("Yoga Mat", 100)

# Invalid quantity test
change_quantity("Yoga Mat", -10)

# d) Sort products
print("\nProducts sorted by quantity:")
print(sort_products("Qty"))

print("\nProducts sorted by price:")
print(sort_products("Price_per_unit"))

print("\nProducts sorted by weight:")
print(sort_products("Weight_unit"))

# e) Shopping cart demonstration
cart = [
    ("Desk Lamp", 2),
    ("Water Filter", 4),
    ("Yoga Mat", 1),
    ("Unknown Item", 3)
]

valid_cart = check_cart(cart)
print("\nValid shopping cart:")
print(valid_cart)

costs = calculate_cart_costs(valid_cart)
print("\nCalculated costs:")
print(costs)

print_invoice(valid_cart, costs)
```

#Problem 2:
Run-length encoding and character frequency analysis
</div>

```python
# 2a) Encode a string

def encode(msg):
    """Encode repeated characters, e.g. AAABB becomes 3A2B."""
    if msg == "":
        return ""

    encoded = ""
    current = msg[0]
    count = 1

    for char in msg[1:]:
        if char == current and count < 9:
            count += 1
        else:
            encoded += str(count) + current
            current = char
            count = 1

    encoded += str(count) + current
    return encoded


# 2b) Get count and character pairs

def pairs(encoded):
    """Yield the count and character from each pair."""
    if len(encoded) % 2 != 0:
        raise ValueError("Encoded string must have an even length.")

    for i in range(0, len(encoded), 2):
        count = int(encoded[i])
        char = encoded[i + 1]
        yield count, char


# 2c) Decode a string

def decode(encoded):
    """Decode a run-length encoded string."""
    message = ""

    for count, char in pairs(encoded):
        message += char * count

    return message


# 2d) Count characters in an encoded string

def char_frequencies(encoded):
    """Return the frequency of each character."""
    freq = {}

    for count, char in pairs(encoded):
        freq[char] = freq.get(char, 0) + count

    return freq


# Test Cases

print(encode("MISSIPPI"))        # example from the question
print(encode("AAABBCBBB"))       # example from the question
 
phrase = "Heard melodies are sweet, but those unheard are sweeter."
 
encoded_phrase = encode(phrase)
print("\nOriginal:", phrase)
print("Encoded: ", encoded_phrase)
 
print("\nPairs   :", list(pairs(encoded_phrase))[:6], "...")
 
decoded_phrase = decode(encoded_phrase)
print("\nDecoded: ", decoded_phrase)
print("decode(encode(msg)) == msg ?", decoded_phrase == phrase)
 
freq = char_frequencies(encoded_phrase)
print("\nCharacter frequencies:")
print(freq)
 
# a self-check: compare with Python's own counting
all_correct = True
for char in freq:
    if freq[char] != phrase.count(char):
        all_correct = False
print("Frequencies agree with str.count ?", all_correct)
print("Total characters:", sum(freq.values()), "=", len(phrase))
 
# extra edge cases
print("\nEdge cases:", repr(encode("")), repr(decode("")), encode("A" * 12))
 
```

#Problem 3: 
Markov chain location probability and reward optimization
</div>



```python
import numpy as np

# Data from the question
P_walk = np.array([
    [0.10, 0.30, 0.20, 0.40],
    [0.25, 0.50, 0.15, 0.10],
    [0.40, 0.02, 0.50, 0.08],
    [0.30, 0.30, 0.10, 0.30]
])

P_bus = np.array([
    [0.00, 0.60, 0.40, 0.00],
    [0.50, 0.00, 0.40, 0.10],
    [0.80, 0.05, 0.00, 0.15],
    [0.55, 0.05, 0.40, 0.00]
])
matrices = {"w": P_walk, "b": P_bus}
rho = np.array([2, 1, 5, 3])
current_state = np.array([1.0, 0.0, 0.0, 0.0])
np.set_printoptions(precision=4, suppress=True)

# Check transition matrices
print("P_walk rows sum to 1:", np.allclose(P_walk.sum(axis=1), 1))
print("P_bus rows sum to 1:", np.allclose(P_bus.sum(axis=1), 1))

# 3a) Distribution after k journeys
def trans(current_state, P, k):
    s = np.array(current_state, dtype=float)

    for _ in range(k):
        s = s @ P
    return s

# 3b) Expected reward after k journeys
def expected_reward(current_state, P, rho, k):
    s = trans(current_state, P, k)
    return rho @ s

# 3c) Total expected reward for an action sequence
def total_expected_reward(tau, current_state, rho):
    s = np.array(current_state, dtype=float)
    total = 0.0

    for action in tau:
        if action not in matrices:
            raise ValueError("Use 'w' for walking or 'b' for bus.")

        s = s @ matrices[action]
        total += rho @ s

    return total

# 3d) Generate random action sequences
## Testing different combinations of walking and bus journeys
def random_sequences(n_sequences=100, length=50):
    sequences = []

    for _ in range(n_sequences):
        tau = np.random.choice(
            ["w", "b"], size=length
        ).tolist()
        sequences.append(tau)

    return sequences

def score_sequences(sequences, current_state, rho):
    rewards = []

    for tau in sequences:
        reward = total_expected_reward(
            tau, current_state, rho
        )
        rewards.append(reward)

    return rewards

def keep_best(sequences, rewards, n_best=10):
    order = np.argsort(rewards)[::-1]
    best = []

    for i in order[:n_best]:
        best.append((rewards[i], sequences[i]))

    return best

def search_best_sequences(
    current_state, rho, n_sequences=100,
    length=50, n_best=10
):
    sequences = random_sequences(n_sequences, length)
    rewards = score_sequences(sequences, current_state, rho)

    return keep_best(sequences, rewards, n_best)

# Test cases
print("\n(a) Distribution after journeys")
print("Walking, k = 1:", trans(current_state, P_walk, 1))
print("Bus, k = 1:", trans(current_state, P_bus, 1))
print("Walking, k = 10:", trans(current_state, P_walk, 10))
print("Bus, k = 10:", trans(current_state, P_bus, 10))

print("\n(b) Expected reward")
print("Walking, k = 1:",
      round(expected_reward(current_state, P_walk, rho, 1), 4))
print("Walking, k = 10:",
      round(expected_reward(current_state, P_walk, rho, 10), 4))
print("Bus, k = 10:",
      round(expected_reward(current_state, P_bus, rho, 10), 4))

print("\n(c) Total expected reward")
print("Sequence w, b, w, w:",
      round(total_expected_reward(
          ["w", "b", "w", "w"], current_state, rho), 4))
print("Walking for 10 journeys:",
      round(total_expected_reward(
          ["w"] * 10, current_state, rho), 4))
print("Bus for 10 journeys:",
      round(total_expected_reward(
          ["b"] * 10, current_state, rho), 4))

print("\n(d) Best 10 of 100 random sequences")
np.random.seed(42)
best = search_best_sequences(current_state, rho)

for rank, (reward, tau) in enumerate(best, start=1):
    print(f"{rank:2}. Reward = {reward:.3f}  {''.join(tau)}")
```

Problem 4: Theater seat booking and reservation system
</div>

```python
# Data from the question

seat_prices = {"P": 120, "D": 45, "*": 75}


# Create the theatre layout
def make_theatre():
    layout = [
        "D***D",
        "*****",
        "D*P*D",
        "*PPP*",
        "DPPPD"
    ]

    theatre = []
    for row in layout:
        theatre.append(list(row))

    return theatre


# Display the seats
def show_theatre(theatre):
    for row in theatre:
        for seat in row:
            print(seat, end=" ")
        print()


# 4a) Find possible seat locations
def find_start_locations(theatre, n_guests, disabled=False):
    locations = []

    if isinstance(n_guests, bool) or not isinstance(n_guests, int) or n_guests < 1:
        return locations

    for row in range(len(theatre)):
        for first_seat in range(len(theatre[row]) - n_guests + 1):
            block = theatre[row][first_seat:first_seat + n_guests]

            if "X" in block:
                continue

            if disabled and "D" not in block:
                continue

            if not disabled and "D" in block:
                continue

            locations.append([row, first_seat])

    return locations


# 4b) Find the most expensive booking
def booking_cost(theatre, row, first_seat, n_guests):
    cost = 0

    for seat in range(first_seat, first_seat + n_guests):
        cost += seat_prices[theatre[row][seat]]

    return cost


def best_booking(theatre, locations, n_guests):
    if not locations:
        return None, 0, []

    costs = []

    for row, first_seat in locations:
        costs.append(booking_cost(theatre, row, first_seat, n_guests))

    best_index = costs.index(max(costs))

    return locations[best_index], costs[best_index], costs


# 4c) Mark the seats as taken
def mark_seats_taken(theatre, row, first_seat, n_guests):
    seats = []

    for seat in range(first_seat, first_seat + n_guests):
        theatre[row][seat] = "X"
        seats.append([row, seat])

    return seats


# 4d) Handle the booking process
def book_party(theatre, n_guests, disabled=False):
    if isinstance(n_guests, bool) or not isinstance(n_guests, int) or n_guests < 1:
        print("Booking failed: enter a valid number of guests.")
        return None

    if not isinstance(disabled, bool):
        print("Booking failed: disabled must be True or False.")
        return None

    try:
        locations = find_start_locations(theatre, n_guests, disabled)

        if not locations:
            print("Sorry, no suitable seats are available.")
            return None

        location, cost, costs = best_booking(theatre, locations, n_guests)

        row, first_seat = location
        seats = mark_seats_taken(theatre, row, first_seat, n_guests)

        print("Booked {} seat(s) in row {}, starting at seat {}. Bill: £{}".format(
            n_guests, row, first_seat, cost
        ))

        return {"seats": seats, "bill": cost}

    except (IndexError, KeyError, TypeError, ValueError) as error:
        print("Booking failed: {}".format(error))
        return None


# Test cases
print("Initial theatre layout:")
theatre = make_theatre()
show_theatre(theatre)

print("\nPossible locations for a party of 2:")
locations = find_start_locations(theatre, 2)
print(locations)

location, cost, costs = best_booking(theatre, locations, 2)
print("Booking costs:", costs)
print("Most expensive booking:", location, "Cost: £" + str(cost))

print("\nTesting the booking process:")
theatre = make_theatre()

book_party(theatre, 2, True)
book_party(theatre, 3)
book_party(theatre, 2)
book_party(theatre, 0)

print("\nFinal theatre layout:")
show_theatre(theatre)

print("\nTesting a party that is too large:")
book_party(theatre, 6)

print("\nTesting when no seats are available:")
full_theatre = [["X"] * 5 for _ in range(5)]
book_party(full_theatre, 1, True)
```
